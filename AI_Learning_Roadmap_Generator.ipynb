```python
SYSTEM_PROMPT = """
You are an expert learning-path designer.

Your job is to create realistic, structured learning roadmaps
for people learning technical and non-technical subjects.

You must:
- respect the learner's current skill level
- respect the available learning time
- arrange topics in logical prerequisite order
- avoid unnecessary topics
- include practical projects
- include milestones
- recommend what to learn first
- make the roadmap achievable within the given timeframe

Do not overload beginners with advanced material.
"""
```

```python
def build_prompt(domain, level, time):
    prompt = f"""
Create a learning roadmap for the following learner.

Domain: {domain}
Current skill level: {level}
Available learning time: {time}

Create a practical roadmap divided into phases.

For each phase include:
1. Phase name
2. Topics to learn
3. Important concepts
4. Practical exercises
5. Project
6. Expected outcome

Also include:
- Prerequisites
- Tools/resources to learn
- Final project
- Milestones
- What the learner should be able to do at the end

Make the roadmap realistic for the available time.

Return the answer in clean Markdown.
"""
    return prompt
```

```python
def generate_roadmap(domain, level, time):

    prompt = build_prompt(domain, level, time)

    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {
                "role": "system",
                "content": SYSTEM_PROMPT
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.7,
    )

    return response.choices[0].message.content
```

```python
roadmap = generate_roadmap(
    "Machine Learning",
    "Beginner",
    "3 months"
)

print(roadmap)
```

```python
import gradio as gr

with gr.Blocks() as demo:

    gr.Markdown(
        """
        # 🎓 AI Learning Roadmap Generator

        Create a personalized learning roadmap using AI.
        """
    )

    domain = gr.Textbox(
        label="Domain / Field",
        placeholder="e.g. Machine Learning"
    )

    level = gr.Dropdown(
        choices=[
            "Beginner",
            "Intermediate",
            "Advanced"
        ],
        label="Current Skill Level"
    )

    time = gr.Textbox(
        label="Learning Time",
        placeholder="e.g. 3 months"
    )

    generate_btn = gr.Button("Generate Roadmap")

    output = gr.Markdown(
        label="Your Roadmap"
    )

    generate_btn.click(
        fn=generate_roadmap,
        inputs=[domain, level, time],
        outputs=output
    )
```
